## Hypothesis


We hypothesize that the relationship between San Diego Trolley ridership and nearby crime rates over time will be weak or mixed, rather than uniformly positive or negative. This is because we expect that any observed differences in crime patterns will be driven largely by preexisting neighborhood socioeconomic conditions, rather than the transit use itself. This aligns with prior research suggesting public transit is often blamed for crime patterns that reflect broader structural inequities rather than transit use.

Null Hypothesis: Changes in San Diego Trolley ridership have no association with nearby crime rates over time.

Alternative Hypothesis: Any association between trolley ridership and crime varies across neighborhoods and is more strongly related to socioeconomic context than to ridership levels alone.

```python
# Drop any rows or columns are completely null
clean_transit_df = transit_df.dropna(how='all', axis=0)
clean_transit_df = transit_df.dropna(how='all', axis=1)
```

```python
# Shape of dataset
print(f'Columns in dataset: {clean_transit_df.shape[1]}\nRows in dataset: {clean_transit_df.shape[0]}')
```

```python
# Seeing what percent of columns have null values
empty_columns = list()
for col in clean_transit_df.columns:
    if ((clean_transit_df.shape[0] - clean_transit_df[col].count()) / clean_transit_df.shape[0]) * 100 > 5:
        empty_columns.append(col)
        
    print(f'{col}: {round(((clean_transit_df.shape[0] - clean_transit_df[col].count()) / clean_transit_df.shape[0]) * 100, 2)}% null\n----------------------')
```

```python
# Focusing on columns which have null values over 5% (I will call them empty_columns)

for col in empty_columns:
    print(f'Different kinds of values in {col}:')
    print(clean_transit_df[col].value_counts())
    print('-----------------')
    


```

```python
# The columns look to either be irrelevant to our project or too imcomplete to evaluate
# So lets drop them

print('Dropping empty_columns')
clean_transit_df = clean_transit_df.drop(columns=empty_columns)
clean_transit_df
```

```python
# Looks like stop_code imitates stop_id whenever it is an integer. So we can drop stop_code

print('Dropping "stop_code"')
clean_transit_df = clean_transit_df.drop(columns='stop_code')
clean_transit_df
```

```python
# Looks like "stop_uid" is a combination of "stop_id" and "stop_agncy" so we can drop it

print('Dropping "stop_uid"')
clean_transit_df = clean_transit_df.drop(columns='stop_uid')
clean_transit_df
```

```python
# lets take a closer look at the differences between stop_lat/lon and lat/lng

coord_df = clean_transit_df.copy()
coord_df['lat_diff'] = round(coord_df['stop_lat'] - coord_df['lat'], 6)
coord_df['lon_diff'] = round(coord_df['stop_lon'] - coord_df['lng'], 6)
print(coord_df['lat_diff'].value_counts())
print(coord_df['lon_diff'].value_counts())

print('\n The difference is very minimal and uniform so we can use stop_lat/lon going forward')
clean_transit_df = clean_transit_df.drop(columns=['lat','lng'])
clean_transit_df
```

```python
print('Now we will load the new clean_transit_df data into "data/02-processed"')

import os
folder_name = 'Group022_WI26/data/02-processed'
file_name = 'clean_transit_stops_datasd.csv'
full_path = os.path.join(folder_name, file_name)

# Make sure folder exists
os.makedirs(folder_name, exist_ok=True)

# Save the CSV in the correct folder
clean_transit_df.to_csv(full_path, index=False)

print(f"Data saved to {full_path}")
```
