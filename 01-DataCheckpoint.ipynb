## Research Question

Is there an association between increased ridership at San Diego Trolley stations and nearby crime rates over time? Are these effects different across neighborhood socioeconomic contexts?


## Background and Prior Work

Many people argue against the implementation of public transportation in their neighborhoods, citing its tendency to bring crime into their communities. But is this really true? Much research has been done looking into the relationship between public transportation and crime rates and different conclusions have been drawn. Some claim that public transportation can “import” crime into wealthy neighborhoods, leading to higher than normal crime rates. Others see this crime “import” effect as unconvincing, and instead find public transportation alleviates socioeconomic stressors which provoke people into doing crime in the first place.

An issue presented when studying public transportation, particularly discerning its impact on crime rates, is the seeming lack of A/B testing. A place that has access to public transportation and one that doesn’t cannot typically be tested against each other without accounting for a multitude of external factors. This makes sense because many things contribute to whether or not a place is given the opportunity to have public transportation or not. In a clever attempt to navigate this issue, David C. Phillips and Danielle Sandler<a name="cite_note-1"></a> [^](#cite_ref-1) wrote a paper analyzing crime rates in the Washington D.C. area when certain transit stations were closed. The paper found that crime decreases across the railway network when stations are closed and that the decrease in crime was felt more in areas where fewer offenders lived. In essence, the closure of stations can inhibit how easily criminals can navigate the network.

Another project, completed by Greg Ridgeway and John MacDonald, looked more holistically at the railway network in Los Angeles over a 27 year span from 1988-2014.<a name="cite_note-2"></a> [^](#cite_ref-2) They conclude that there is little evidence that access to public railways lead to an increase in crime in neighborhoods. While they do concede that criminals could use railways to travel to lower crime, wealthy areas to find victims, they argue that access to public transportation could lower overall crime rates in a metropolitan area. They claim that giving public transit access to people who hail from low socioeconomic areas can help them seek better opportunities rather than resort to crime.

Looking through the lens of public safety, Tanya Lewis writes an article detailing how public transportation results in far fewer deaths and fewer crimes than automobile travel. <a name="cite_note-3"></a> [^](#cite_ref-3) Per 1 billion passenger miles traveled in 2022, automobiles caused 16.8 fatalities in rural areas and 11.5 in urban areas. To contrast, rail transit fatalities were 5.9 per 1 billion miles and many were non-occupant fatalities that resulted from suicide. Additionally, crimes like assault, robbery, and homicide are down on public transportation in 2023 compared to in 2019 while motor vehicle theft and road rage incidents involving guns are up. The article suggests that many people feel more unsafe on public transportation than they should according to the statistics.

Diving deeper into this phenomenon, Anastasia Loukaitou-Sideris and Camille Fink address the fear women feel when using public transportation.<a name="cite_note-4"></a> [^](#cite_ref-4) Women will often avoid certain railways, only use them in the day, or forego public transportation altogether due to fear of victimization. According to Loukaitou-Sideris and Fink, much of this blame should be put into the lack of input women get when designing safety procedures for public transportation. Citing the lack of security, cameras, panic buttons, or other bystanders who could help, the authors argue that women will continue to avoid public transportation so long as there is no greater attempt to prevent crime around transportation infrastructure like bus stops, trains etc. If public transportation does in fact have a way of bringing crime to certain areas, increased security around transit stops could potentially help to mitigate this.

## Hypothesis


We hypothesize that the relationship between San Diego Trolley ridership and nearby crime rates over time will be weak or mixed, rather than uniformly positive or negative. This is because we expect that any observed differences in crime patterns will be driven largely by preexisting neighborhood socioeconomic conditions, rather than the transit use itself. This aligns with prior research suggesting public transit is often blamed for crime patterns that reflect broader structural inequities rather than transit use.

Null Hypothesis: Changes in San Diego Trolley ridership have no association with nearby crime rates over time.

Alternative Hypothesis: Any association between trolley ridership and crime varies across neighborhoods and is more strongly related to socioeconomic context than to ridership levels alone.

### Dataset #1: Transit Stops Attribute Data

The Transit Stops dataset contains geographic and identifying information for public transit stops across the San Diego region (MTS and NCTD). The dataset includes 6,220 observations and 14 variables, with each row representing one transit stop. 的The most important quantitative metrics in the Transit Stops dataset are the spatial coordinates stop_lat and stop_lon (also listed as LAT and LNG for more decimal places).These variables record the geographic position of each transit stop using decimal degree units. Latitude measures north–south position relative to the equator, while longitude measures east–west position relative to the prime meridian. In this dataset, latitude values range roughly from 32.5 to 33.0 degrees, and longitude values range from approximately –117.3 to –116.9 degrees, which corresponds to the broader San Diego County region. These coordinate metrics allow us to map stop locations, calculate spatial distances, and maybe infer correlations, making them foundational for spatial crime analysis.

Another operational metric is the wheelchair, which is recorded as a categorical numeric indicator rather than a continuous measure. The wheelchair accessibility variable follows the General Transit Feed Specification (GTFS) standard, where 0 indicates no accessibility information, 1 indicates wheelchair-accessible boarding, and 2 indicates non-accessible stops. In this dataset, values appear as “0E-8,” which is a scientific notation formatting issue but functionally corresponds to 0 (unknown accessibility status). This metric reflects infrastructure accessibility and may indicate variation in station investment or design equity.

Other identifiers such as stop_id and stop_code function as nominal variables used to identify the stop. They uniquely identify each stop but have no units or ordinal or numerical meaning. 

One major concern with this dataset is that it is not real-time or recorded with time. The data provide a static snapshot of where transit stops are located, but they do not include any information about ridership levels, service frequency, or how ridership changes over time. Because of this, the dataset cannot capture actual transit activity or passenger volume. For our research question, this is important because crime patterns are more likely related to how many people are using a station rather than simply whether a stop exists at that location.

Another limitation is the lack of neighborhood contextual information. The dataset does not include demographic, socioeconomic, or land-use variables for the areas surrounding each stop. Transit stops are often located in neighborhoods with different income levels, population densities, and built environments, all of which may influence crime rates independently of transit use. Without incorporating these contextual factors, it is difficult to determine whether any observed crime patterns are associated with transit ridership or with preexisting neighborhood conditions.

### Dataset #2: 2025 Crime Rates Data

This crime dataset reports crime rates for neighborhoods in San Diego throughout 2025. There are 125 observations, each being a SD neighborhood, and 23 variables, each being a different offense. Besides the Neighborhood column (the unique ID), the data are on the same unit scale of crime per 1,000 residents for the neighborhood. This specific metric allows us to compare crime across neighborhoods and crime “density” to trolley activity, regardless of each neighborhood’s population. Among these different crime offense rates, the most important measures for our research are `[Simple Assault, Intimidation, Robbery, Larceny, Theft From Motor Vehicle, Motor Vehicle Theft, Vandalism, Drug/Narcotic Violations]`. These crimes are intuitively related to public transit because they often occur during commutes or at transit stops, where high foot traffic and accessibility provide opportunities for criminal activity.

To understand the specific scale of crime per 1,000 residents, we can look at the larceny crime rates—the most common public- and transit- related crime. The average “larceny” rate is 15.74, which means that for every 1,000 people, we expect roughly 16 offenses were reported in 2025. Similarly, a rate of 7.88 for vandalism means for every 1,000 people, there were about 8 reported offenses.

Although the data is normalized per 1,000 residents, we cannot easily say that high crime rates indicate “high crime neighborhoods" because these rates often reflect crimes from any person against the permanent resident population. Therefore, we must consider if these neighborhoods have high traffic or massive tourism like Mission Bay, or serve as a central connection hub for many public transit services like Old Town. Ultimately, the influx of people artificially inflates the crime rates relative to the number of residents. However, taking this into consideration, we can use this to our advantage to infer if inflation is related to trolley operations.

Like Dataset #1, one major concern with this dataset is the lack of temporal granularity. Since it only provides aggregated crime rates for the year 2025, the data do not distinguish between crimes that occur during peak hours—high ridership—or during low activity hours. This makes it difficult to definitively prove a causal relationship between transit activities and crime without additional information on each crime incident.

Another concern with this dataset is that it only includes crimes that were officially reported to and recorded by the SD Police Department. Many incidents go unreported for multiple reasons, including no or delayed responses by police or distrust of law enforcement. Additionally, there is potential for false police reports in this data that may affect  our analysis of socioeconomic factors.
