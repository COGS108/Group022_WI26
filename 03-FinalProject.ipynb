## Research Question

Is there an association between increased ridership at San Diego Trolley stations and nearby crime rates over time? Are these effects different across neighborhood socioeconomic contexts?

## Background and Prior Work

Many people argue against the implementation of public transportation in their neighborhoods, citing its tendency to bring crime into their communities. But is this really true? Much research has been done looking into the relationship between public transportation and crime rates and different conclusions have been drawn. Some claim that public transportation can “import” crime into wealthy neighborhoods, leading to higher than normal crime rates. Others see this crime “import” effect as unconvincing, and instead find public transportation alleviates socioeconomic stressors which provoke people into doing crime in the first place.

An issue presented when studying public transportation, particularly discerning its impact on crime rates, is the seeming lack of A/B testing. A place that has access to public transportation and one that doesn’t cannot typically be tested against each other without accounting for a multitude of external factors. This makes sense because many things contribute to whether or not a place is given the opportunity to have public transportation or not. In a clever attempt to navigate this issue, David C. Phillips and Danielle Sandler<a name="cite_note-1"></a> [^](#cite_ref-1) wrote a paper analyzing crime rates in the Washington D.C. area when certain transit stations were closed. The paper found that crime decreases across the railway network when stations are closed and that the decrease in crime was felt more in areas where fewer offenders lived. In essence, the closure of stations can inhibit how easily criminals can navigate the network.

Another project, completed by Greg Ridgeway and John MacDonald, looked more holistically at the railway network in Los Angeles over a 27 year span from 1988-2014.<a name="cite_note-2"></a> [^](#cite_ref-2) They conclude that there is little evidence that access to public railways lead to an increase in crime in neighborhoods. While they do concede that criminals could use railways to travel to lower crime, wealthy areas to find victims, they argue that access to public transportation could lower overall crime rates in a metropolitan area. They claim that giving public transit access to people who hail from low socioeconomic areas can help them seek better opportunities rather than resort to crime.

Looking through the lens of public safety, Tanya Lewis writes an article detailing how public transportation results in far fewer deaths and fewer crimes than automobile travel. <a name="cite_note-3"></a> [^](#cite_ref-3) Per 1 billion passenger miles traveled in 2022, automobiles caused 16.8 fatalities in rural areas and 11.5 in urban areas. To contrast, rail transit fatalities were 5.9 per 1 billion miles and many were non-occupant fatalities that resulted from suicide. Additionally, crimes like assault, robbery, and homicide are down on public transportation in 2023 compared to in 2019 while motor vehicle theft and road rage incidents involving guns are up. The article suggests that many people feel more unsafe on public transportation than they should according to the statistics.

Diving deeper into this phenomenon, Anastasia Loukaitou-Sideris and Camille Fink address the fear women feel when using public transportation.<a name="cite_note-4"></a> [^](#cite_ref-4) Women will often avoid certain railways, only use them in the day, or forego public transportation altogether due to fear of victimization. According to Loukaitou-Sideris and Fink, much of this blame should be put into the lack of input women get when designing safety procedures for public transportation. Citing the lack of security, cameras, panic buttons, or other bystanders who could help, the authors argue that women will continue to avoid public transportation so long as there is no greater attempt to prevent crime around transportation infrastructure like bus stops, trains etc. If public transportation does in fact have a way of bringing crime to certain areas, increased security around transit stops could potentially help to mitigate this.

## Hypothesis


We hypothesize that the relationship between San Diego Trolley ridership and nearby crime rates over time will be weak or mixed, rather than uniformly positive or negative. This is because we expect that any observed differences in crime patterns will be driven largely by preexisting neighborhood socioeconomic conditions, rather than the transit use itself. This aligns with prior research suggesting public transit is often blamed for crime patterns that reflect broader structural inequities rather than transit use.

Null Hypothesis: Changes in San Diego Trolley ridership have no association with nearby crime rates over time.

Alternative Hypothesis: Any association between trolley ridership and crime varies across neighborhoods and is more strongly related to socioeconomic context than to ridership levels alone.

### Merging all datasets
Now that we have our crime rate dataset all together, we can start merging everything together.

```python
# Importing all 4 processed datasets

clean_stops_df = pd.read_csv('data/02-processed/clean_transit_stops_datasd.csv')
clean_crime_df = pd.read_csv('data/02-processed/clean_neighborhood_common_crime_rates.csv')
clean_ridership_df = pd.read_csv('data/02-processed/cleaned_mts_trolley_ridership_data.csv')
clean_equity_df = pd.read_csv("data/02-processed/clean_equity.csv")
```

```python
# Reformats naming convention of 'Station' in our transit stops data to match that of ridership data

def clean_stops(t):
    output = re.sub(r'[^\w\s]', ' ', t).lower().replace('station', '')
    output = re.sub(r'\bstreet\b|\bst\b', '', output, flags=re.IGNORECASE)
    output = re.sub(r'\bavenue\b|\bave\b', '', output, flags=re.IGNORECASE)
    return re.sub(r'\s+', ' ', output).strip()
```

```python
# Creates a new transit stops column following clean station name formatting

clean_ridership_df['stop_join'] = clean_ridership_df['Station'].apply(clean_stops)

clean_stops_df['stop_join'] = np.where(
    (clean_stops_df['stop_id'].str[:2] == '75') | (clean_stops_df['stop_id'].str[:2] == '77'),
    clean_stops_df['stop_name'].apply(clean_stops), 
    np.nan
)
```

```python
clean_stops_df.loc[~clean_stops_df['stop_join'].isin(clean_ridership_df['stop_join']), 'stop_join'].unique()
```

```python
clean_ridership_df.loc[clean_ridership_df['stop_join'].str.match('santee|grantville'), 'stop_join'].unique()
```

```python
# Resolves a naming conflict between transit stop and ridership dataset

clean_stops_df['stop_join'] = clean_stops_df['stop_join'].str.replace('santee town center', 'santee').str.replace('union grantville', 'grantville')
```

```python
# Groups transit stop data by newly formatted station name and takes mean of each coordinate to resolve any slight discrepancies

clean_stops_df = clean_stops_df.groupby('stop_join').agg({'stop_lat': 'mean', 'stop_lon': 'mean'}).reset_index()
clean_stops_df.head()
```

```python
#  Merges transit stop and ridership data on universally formatted station names

transit_crime_df = clean_stops_df.merge(clean_ridership_df, on='stop_join').drop(columns=['stop_join'])
transit_crime_df.head()
```

```python
# Adds crime data for the neighborhood the transit stop resides in

transit_crime_df = transit_crime_df.merge(clean_crime_df, on=['Neighborhood', 'Year'])
transit_crime_df.head()
```

```python
# Since there are many indicators and many subcategories, we will keep a subset of columns to avoid introducing any noise. 
# Additionally, we found that many indicator subcategories have no meaningful shifts in percentage or remained constant over time. 
# Therefore, we priortize categories with measurable shifts and treat those without any shifts as baseline constants.

prevalent_indicators = clean_equity_df['indicator'].unique()
prevalent_cat = clean_equity_df['subcategory'].unique()

prevalent_indicators = clean_equity_df.copy()

prevalent_indicators['indicator_subcategory'] = prevalent_indicators['indicator'] + '_' + prevalent_indicators['subcategory']
equity_pivot_table = prevalent_indicators.pivot_table(index='equity_report_year', columns='indicator_subcategory', values='percent').reset_index()
```

```python
# Checking for significant shifts
equity_shifts = equity_pivot_table.iloc[:, 1:].diff().iloc[1]
equity_shifts = equity_shifts[equity_shifts.abs() > 0].sort_values(ascending=False).round(1)
display_n = 10
equity_shifts.name = '2023-2025 Equity Shifts'
pd.concat([equity_shifts[:display_n], equity_shifts[-display_n:]])
```

```python
# To guarantee this variable selection is accurate, we check the raw counts of specific indicators to ensure these percent shifts are meaningful rather than data errors. 
# As examples, let's check the highest percent shift and lowest percent shift.

clean_equity_df.loc[(clean_equity_df['indicator'] == 'Self-Sufficiency Wage') & (clean_equity_df['subcategory'] == 'Yes'), ['equity_report_year', 'subcategory', 'count', 'percent']]
```

```python
clean_equity_df.loc[(clean_equity_df['indicator'] == 'High School Graduates') & (clean_equity_df['subcategory'] == 'Some Other Race'), ['equity_report_year', 'subcategory', 'count', 'percent']]

# The largest negative shift appears to be reporting error.
```

```python
# Final selection of equity indicators
equity_pivot_table = equity_pivot_table[[
    'equity_report_year', 'Self-Sufficiency Wage_Yes', 
    'Business Ownership_Minority', 'Commute Time and Method of Transportation_Total Private Motor Vehicle', 
    'English Language Learners_Level 1'
]]
equity_pivot_table.info()
equity_pivot_table.head()
```

```python
# Final merge
df = transit_crime_df.merge(equity_pivot_table, left_on='Year', right_on='equity_report_year').drop(columns=['equity_report_year'])
df.info()
df.head()
```

```python
folder_name = os.path.join('data', '02-processed')
file_name = f'final_transit_crime_data_cleaned.csv'
full_path = os.path.join(folder_name, file_name)
os.makedirs(folder_name, exist_ok=True)
df.to_csv(full_path, index=False)
print(f'Data successfully saved to "{full_path}"')
```

Before we begin modeling, we collapse the data from line to station. While exploding the data based on line let us simplify and easily interpret our EDA, this complicates statistical methods.

```python
# Reset df
df = pd.read_csv('data/02-processed/final_transit_crime_data_cleaned.csv')

# Collapse dataset from line to station
df = df.groupby(['Station', 'Year', 'Season', 'Neighborhood'])[df.drop(columns=['Station', 'Year', 'Season', 'Neighborhood', 'Line']).columns].mean().reset_index()
df['Total Crime'] = df[crime_cols].sum(axis=1)
df.head()
```

```python
annual_crime = df.groupby(['Year', 'Station'])['Total Crime'].mean().reset_index()
annual_crime_df = annual_crime.pivot(index='Station', columns='Year', values='Total Crime').dropna()
ttest_crime = stats.ttest_rel(annual_crime_df[2023], annual_crime_df[2025])
print(ttest_crime)
```

```python
annual_ridership = df.groupby(['Year', 'Station'])['Ridership'].mean().reset_index()
annual_ridership_df = annual_ridership.pivot(index='Station', columns='Year', values='Ridership').dropna()
ttest_rider = stats.ttest_rel(annual_ridership_df[2023], annual_ridership_df[2025])
print(ttest_rider)
```

```python
diff = annual_ridership_df[2025] - annual_ridership_df[2023]
cohens_d = np.mean(diff) / np.std(diff, ddof=1)
print(f"Cohen's d: {cohens_d:.4f}")
```

```python
for c in crime_cols:
    frm = f'Q("{c}") ~ Ridership * C(Year)'
    mdl = smf.ols(frm, data=df).fit()
    coeffs = pd.DataFrame(mdl.params).rename(columns={0:f'coefficient'})
    pvals = pd.DataFrame(mdl.pvalues).rename(columns={0:f'p-value'})
    print(f'{c} (R-squared={mdl.rsquared:.3f})')
    display(pd.concat([coeffs, pvals], axis=1).round(5))
    print()
```
